# Assignment 6: Filtering DataFrames

I need some quick research on store 25:

* First, calculate the percentage of times ALL stores had more than 2000 transactions
* Then, calculate the percentage of times store 25 had more than 2000 transactions, and calculate the sum of transactions on these days
* Finally, sum the transactions for stores 25 and 31, that occurred in May or June, and had less than 2000 transactions


```python
df.head()
```

```python
daily_total = df.groupby('date')['transactions'].sum()
days_above_2000 = (daily_total > 2000).sum()
total_days = df['date'].nunique()
percentage = (days_above_2000 / total_days) * 100
print(f"Percentage of days when total transactions were > 2000: {percentage:.2f}%")
```

```python
store_25_df = df[df['store_nbr'] == 25]
mask_above_2000 = store_25_df['transactions'] > 2000
days_more_2000 = mask_above_2000.sum()

total_days_25_store = len(store_25_df)
percentage_25 = (days_more_2000 / total_days_25_store) * 100 
sum_transactions_25 = store_25_df.loc[mask_above_2000, 'transactions'].sum()

print(f"Store 25 had more than 2000 transactions on {percentage_25:.2f}% of its days.")
print(f"The total sum of transactions on those days is: {sum_transactions_25:,}")
```

```python
mask_stors = df['store_nbr'].isin([25, 31])
mask_months = df['date'].dt.month.isin([5, 7])
mask_transactions = df['transactions'] > 2000
final_mask = mask_stors & mask_months & mask_transactions
sum_transactions = df.loc[final_mask, 'transactions'].sum()

print(f"The sum of transactions for the specified criteria is: {sum_transactions:,.2f}")
```
