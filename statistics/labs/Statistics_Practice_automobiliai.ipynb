Kaip pasiskirsčiusios kuro sąnaudos? pavaizduokite grafiškai

```python
plt.figure(figsize=(10, 6))
ax = sns.histplot(df["kuro_sanaudos"], kde=True, color="green", bins=20)
mean_val = df["kuro_sanaudos"].mean()
plt.axvline(mean_val, color='red', linestyle='--', label=f'Mean: {mean_val:.2f}')

plt.title('Fuel Consumption')
plt.xlabel('Fuel consumption (l/100km)')
plt.ylabel('Number of cars')
plt.legend()
plt.show()
```

Kaip pasiskirsčiusios Arklio galios? pavaizduokite grafiškai

```python
plt.figure(figsize=(10, 6))
ax = sns.histplot(df["arklio_galios"], kde=True, color="blue", bins=20)
mean_val = df["arklio_galios"].mean()
plt.axvline(mean_val, color='red', linestyle='--', label=f'Mean: {mean_val:.2f} HP')

plt.title('Horsepower Distribution')
plt.xlabel('Horsepower (HP)')
plt.ylabel('Number of engines')
plt.legend()
plt.show()
```
